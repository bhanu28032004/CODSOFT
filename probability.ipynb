```python
# CODE BLOCK 2: Step 2 - Random Line Selection

print("\n--- Step 2: Random Line Selection ---")

# Randomly select a dimension (x or y)
selected_axis = random.choice(['x', 'y']) # This needs to be a global var or passed

if selected_axis == 'y': # Vertical line: x = random_x_coord
    random_coord = random.uniform(face_min_x, face_max_x) # This needs to be a global var or passed
    line_x_coords = [random_coord, random_coord]
    line_y_coords = [0, 1] # Extend across the whole plot
    line_label = f"Vertical line: x = {random_coord:.2f}"
    print(f"Selected axis for the line: X-axis")
    print(f"Randomly selected x-coordinate for the vertical line: x = {random_coord:.2f}")
else: # Horizontal line: y = random_y_coord
    random_coord = random.uniform(face_min_y, face_max_y) # This needs to be a global var or passed
    line_x_coords = [0, 1] # Extend across the whole plot
    line_y_coords = [random_coord, random_coord]
    line_label = f"Horizontal line: y = {random_coord:.2f}"
    print(f"Selected axis for the line: Y-axis")
    print(f"Randomly selected y-coordinate for the horizontal line: y = {random_coord:.2f}")

# Create a new figure and axis for this step's output
fig2, ax2 = plt.subplots(figsize=(7,7))
create_smiley_face_plot_base(ax2) # Draw the base smiley face
ax2.plot(line_x_coords, line_y_coords, color='purple', linestyle='--', linewidth=3, zorder=7, label=line_label)
ax2.set_title('Step 2: Random Line Selection', fontsize=14)
ax2.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

```python
# CODE BLOCK 3: Step 3 - Select Random Points on Line within Face Area

print("\n--- Step 3: Selecting Random Points on Line within Face Area ---")

# Define the face's circular boundary for point validation
# Center (0.5, 0.5), Radius 0.39
face_center_x, face_center_y = 0.5, 0.5
face_radius = 0.39

selected_points_on_line = [] # This will store (x, y) tuples of selected points
num_points_to_select = 10 # Minimum number of points

# Helper function to check if a point is within the circular face
def is_within_face(px, py, center_x, center_y, radius):
    return (px - center_x)**2 + (py - center_y)**2 <= radius**2

# If a vertical line was selected in Step 2 (x = random_coord)
if selected_axis == 'y':
    # The x-coordinate is fixed (random_coord from Step 2)
    # We need to find the min/max y-coordinates on the line that are within the face
    valid_y_coords = []
    # Iterate through possible y-coordinates along the line
    test_y = np.linspace(0, 1, 500) # Fine resolution for testing
    for ty in test_y:
        if is_within_face(random_coord, ty, face_center_x, face_center_y, face_radius):
            valid_y_coords.append(ty)

    if not valid_y_coords:
        print("No valid y-coordinates found on the vertical line within the face.")
    else:
        min_valid_y = min(valid_y_coords)
        max_valid_y = max(valid_y_coords)

        # Select random y-coordinates within this valid range
        for _ in range(num_points_to_select):
            ry = random.uniform(min_valid_y, max_valid_y)
            selected_points_on_line.append((random_coord, ry))

# If a horizontal line was selected in Step 2 (y = random_coord)
else: # selected_axis == 'x'
    # The y-coordinate is fixed (random_coord from Step 2)
    # We need to find the min/max x-coordinates on the line that are within the face
    valid_x_coords = []
    # Iterate through possible x-coordinates along the line
    test_x = np.linspace(0, 1, 500) # Fine resolution for testing
    for tx in test_x:
        if is_within_face(tx, random_coord, face_center_x, face_center_y, face_radius):
            valid_x_coords.append(tx)

    if not valid_x_coords:
        print("No valid x-coordinates found on the horizontal line within the face.")
    else:
        min_valid_x = min(valid_x_coords)
        max_valid_x = max(valid_x_coords)

        # Select random x-coordinates within this valid range
        for _ in range(num_points_to_select):
            rx = random.uniform(min_valid_x, max_valid_x)
            selected_points_on_line.append((rx, random_coord))


# Extract x and y coordinates for plotting
plot_x_points = [p[0] for p in selected_points_on_line]
plot_y_points = [p[1] for p in selected_points_on_line]

print(f"Selected {len(selected_points_on_line)} points on the line within the face area.")

# Create a new figure and axis for this step's output
fig3, ax3 = plt.subplots(figsize=(7,7))
# Assuming create_smiley_face_plot_base(ax) and line_x_coords, line_y_coords, line_label are available from Step 2's context
# You'll need to define create_smiley_face_plot_base(ax) to draw the base smiley face on a given axis.
# For example:
def create_smiley_face_plot_base(ax):
    # This should be the content of your create_smiley_face_plot function, but without creating a new fig, ax
    # Draw the smiley face components
    face = plt.Circle((0.5, 0.5), 0.39, color='#FFD700', ec='black', lw=2, zorder=3)
    ax.add_artist(face)

    eye_left_sclera = plt.Circle((0.36, 0.62), 0.06, color='white', ec='black', lw=2, zorder=4)
    eye_right_sclera = plt.Circle((0.64, 0.62), 0.06, color='white', ec='black', lw=2, zorder=4)
    ax.add_artist(eye_left_sclera)
    ax.add_artist(eye_right_sclera)

    eye_left_pupil = plt.Circle((0.36, 0.62), 0.03, color='black', zorder=5)
    eye_right_pupil = plt.Circle((0.64, 0.62), 0.03, color='black', zorder=5)
    ax.add_artist(eye_left_pupil)
    ax.add_artist(eye_right_pupil)

    blush_left = plt.Circle((0.30, 0.46), 0.03, color='#FFB6C1', zorder=4)
    blush_right = plt.Circle((0.70, 0.46), 0.03, color='#FFB6C1', zorder=4)
    ax.add_artist(blush_left)
    ax.add_artist(blush_right)

    ax.plot([0.32, 0.40], [0.70, 0.74], color='black', lw=2, zorder=6) # Eyebrow left
    ax.plot([0.60, 0.68], [0.74, 0.70], color='black', lw=2, zorder=6) # Eyebrow right

    theta = np.linspace(-np.pi + np.pi/7, -np.pi/7, 100)
    x_smile = 0.5 + 0.20 * np.cos(theta)
    y_smile = 0.40 + 0.13 * np.sin(theta)
    ax.plot(x_smile, y_smile, color='black', lw=4, zorder=6) # Mouth

    ax.plot([0.5, 0.50], [0.53, 0.45], color='black', lw=2, zorder=6) # Nose

    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.set_aspect('equal')
    ax.set_xticks(np.arange(0, 1.05, 0.05))
    ax.set_yticks(np.arange(0, 1.05, 0.05))
    ax.set_xticks(np.arange(0, 1.01, 0.01), minor=True)
    ax.set_yticks(np.arange(0, 1.01, 0.01), minor=True)
    ax.grid(which='major', color='green', linestyle='-', linewidth=0.7)
    ax.grid(which='minor', color='lightgreen', linestyle='--', linewidth=0.3)


create_smiley_face_plot_base(ax3) # Draw the base smiley face
# Redraw the line from Step 2
ax3.plot(line_x_coords, line_y_coords, color='purple', linestyle='--', linewidth=3, zorder=7, label=line_label)
# Plot the newly selected points
ax3.scatter(plot_x_points, plot_y_points, color='red', s=100, zorder=8, label=f'Selected Points (={len(selected_points_on_line)})')

ax3.set_title('Step 3: Selected Random Points on Line within Face Area', fontsize=14)
ax3.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

```python
# CODE BLOCK 4: Step 4 - Centroid Selection and Initial Cluster Assignment (No Image Rendering)

import random
import numpy as np # Used for potential future calculations, but not strictly for this step's output

print("\n--- Step 4: Centroid Selection and Initial Cluster Assignment ---")

# All points from Step 1 (conceptual regions)
all_points = [
    -8.5, -7.8, -8.2,  # Around -8
    -3.2, -2.7, -3.5,  # Around -3
    0.3, 0.6, 0.4,     # Around 0.5
    3.8, 4.2, 4.1,     # Around 4
    7.9, 8.3, 7.7      # Around 8
]

# 4a. Randomly select 2 points as initial centroids
# Make a copy to select without replacement
points_for_centroid_selection = list(all_points)
random.shuffle(points_for_centroid_selection)

centroid1 = points_for_centroid_selection.pop() # Selects and removes one point
centroid2 = points_for_centroid_selection.pop() # Selects and removes another point

# Ensure centroids are distinct, if by chance they were the same (highly unlikely with pop)
# In a real K-Means, you'd handle this, but for this example, we'll assume distinct
if centroid1 == centroid2:
    print("Warning: Centroids are the same. Re-selecting Centroid 2.")
    # This might fail if points_for_centroid_selection is empty, but for 15 points, it's safe.
    centroid2 = points_for_centroid_selection.pop()

print(f"Randomly selected Centroid 1 (C1): {centroid1}")
print(f"Randomly selected Centroid 2 (C2): {centroid2}")

# 4b. Divide the remaining points into clusters based on proximity to centroids
cluster1_points = []
cluster2_points = []

# Robust way to get remaining points for assignment
temp_points_for_assignment = list(all_points)
# Remove centroids from the list of points to be assigned
try:
    temp_points_for_assignment.remove(centroid1)
    temp_points_for_assignment.remove(centroid2)
except ValueError:
    # This handles cases where a centroid value might not be in the list,
    # or if there were duplicates and only one was removed.
    pass
remaining_points_for_assignment = temp_points_for_assignment


print(f"\nPoints to be assigned to clusters (excluding centroids): {remaining_points_for_assignment}")

for point in remaining_points_for_assignment:
    distance_to_c1 = abs(point - centroid1)
    distance_to_c2 = abs(point - centroid2)

    if distance_to_c1 < distance_to_c2:
        cluster1_points.append(point)
    else: # If distances are equal, it defaults to cluster 2 as per the condition
        cluster2_points.append(point)

print(f"\n--- Cluster Assignment Results ---")
print(f"Centroid 1 (C1): {centroid1}")
print(f"Assigned Cluster 1 points: {cluster1_points}")
print(f"Centroid 2 (C2): {centroid2}")
print(f"Assigned Cluster 2 points: {cluster2_points}")

# Store these cluster points for the next steps (mean and standard deviation calculations)
# In a full script, you might pass these lists to the next functions or define them globally.
# For now, these variables will persist in the current execution context.
```

```python
# CODE BLOCK 5: Step 5 - Calculate Mean for Each Cluster

import numpy as np # numpy is excellent for numerical operations like mean

print("\n--- Step 5: Calculate Mean for Each Cluster ---")

# Assuming 'cluster1_points' and 'cluster2_points' are available from the previous step (Step 4)
# For demonstration, I'll use example values that match the previous step's output.
# In a continuous script, these would be the actual variables from Step 4.
example_cluster1_points = [-8.5, -7.8, -3.2, -2.7, -3.5]
example_cluster2_points = [0.3, 0.6, 0.4, 3.8, 4.2, 7.9, 8.3, 7.7]

# Let's use the actual variables if they exist in the execution scope,
# otherwise fall back to the example for standalone execution.
try:
    current_cluster1_points = cluster1_points
    current_cluster2_points = cluster2_points
except NameError:
    print("Using example cluster points as 'cluster1_points' or 'cluster2_points' not found in scope.")
    current_cluster1_points = example_cluster1_points
    current_cluster2_points = example_cluster2_points


print(f"\nPoints in Cluster 1: {current_cluster1_points}")
print(f"Number of points in Cluster 1 (n1): {len(current_cluster1_points)}")
print(f"\nPoints in Cluster 2: {current_cluster2_points}")
print(f"Number of points in Cluster 2 (n2): {len(current_cluster2_points)}")


# --- Mean Calculation for Cluster 1 (µ1) ---
if current_cluster1_points: # Ensure the list is not empty to avoid division by zero
    sum_cluster1 = sum(current_cluster1_points)
    n1 = len(current_cluster1_points)
    mean_cluster1 = sum_cluster1 / n1
    print("\n--- Calculations for Cluster 1 Mean (µ1) ---")
    print(f"Sum of points in Cluster 1: {sum_cluster1}")
    print(f"Number of points in Cluster 1 (n1): {n1}")
    print(f"µ1 = (Sum of points) / n1 = {sum_cluster1} / {n1} = {mean_cluster1:.2f}")
else:
    mean_cluster1 = None
    print("\nCluster 1 is empty, cannot calculate mean.")

# --- Mean Calculation for Cluster 2 (µ2) ---
if current_cluster2_points: # Ensure the list is not empty
    sum_cluster2 = sum(current_cluster2_points)
    n2 = len(current_cluster2_points)
    mean_cluster2 = sum_cluster2 / n2
    print("\n--- Calculations for Cluster 2 Mean (µ2) ---")
    print(f"Sum of points in Cluster 2: {sum_cluster2}")
    print(f"Number of points in Cluster 2 (n2): {n2}")
    print(f"µ2 = (Sum of points) / n2 = {sum_cluster2} / {n2} = {mean_cluster2:.2f}")
else:
    mean_cluster2 = None
    print("\nCluster 2 is empty, cannot calculate mean.")

print(f"\n--- Final Mean Results ---")
print(f"Mean of Cluster 1 (µ1): {mean_cluster1:.2f}")
print(f"Mean of Cluster 2 (µ2): {mean_cluster2:.2f}")

# You would typically store these new means to potentially update centroids in an iterative K-Means algorithm.
# global_mean_cluster1 = mean_cluster1
# global_mean_cluster2 = mean_cluster2
```

```python
# CODE BLOCK 6: Step 6 - Calculate Standard Deviation for Each Cluster

import numpy as np # numpy is essential for efficient statistical calculations

print("\n--- Step 6: Calculate Standard Deviation for Each Cluster ---")

# Assuming 'current_cluster1_points', 'current_cluster2_points',
# 'mean_cluster1', and 'mean_cluster2' are available from previous steps.
# For demonstration, I'll use example values.
example_cluster1_points = [-8.5, -7.8, -3.2, -2.7, -3.5]
example_mean_cluster1 = -5.14 # From previous calculations
example_cluster2_points = [0.3, 0.6, 0.4, 3.8, 4.2, 7.9, 8.3, 7.7]
example_mean_cluster2 = 4.15 # From previous calculations

# Try to use actual variables from previous steps if available
try:
    current_cluster1_points = cluster1_points
    current_mean_cluster1 = mean_cluster1
    current_cluster2_points = cluster2_points
    current_mean_cluster2 = mean_cluster2
except NameError:
    print("Using example cluster points and means as variables not found in scope.")
    current_cluster1_points = example_cluster1_points
    current_mean_cluster1 = example_mean_cluster1
    current_cluster2_points = example_cluster2_points
    current_mean_cluster2 = example_mean_cluster2

print(f"\nCluster 1 points: {current_cluster1_points}")
print(f"Mean of Cluster 1 (µ1): {current_mean_cluster1:.2f}")
print(f"\nCluster 2 points: {current_cluster2_points}")
print(f"Mean of Cluster 2 (µ2): {current_mean_cluster2:.2f}")


# --- Standard Deviation Calculation for Cluster 1 (σ1) ---
print("\n--- Calculations for Cluster 1 Standard Deviation (σ1) ---")
if current_cluster1_points and current_mean_cluster1 is not None:
    n1 = len(current_cluster1_points)
    if n1 > 1: # Standard deviation requires at least 2 points
        # Calculate sum of squared differences from the mean
        sum_sq_diff_cluster1 = sum([(x - current_mean_cluster1)**2 for x in current_cluster1_points])
        variance_cluster1 = sum_sq_diff_cluster1 / (n1 - 1) # Sample variance (n-1 degrees of freedom)
        std_dev_cluster1 = math.sqrt(variance_cluster1)

        print(f"Sum of (x - µ1)^2 for Cluster 1:")
        for x_val in current_cluster1_points:
            diff_sq = (x_val - current_mean_cluster1)**2
            print(f"  ({x_val:.1f} - {current_mean_cluster1:.2f})^2 = {diff_sq:.2f}")
        print(f"  Sum = {sum_sq_diff_cluster1:.2f}")

        print(f"Number of points in Cluster 1 (n1): {n1}")
        print(f"Variance (σ1^2) = Sum((x - µ1)^2) / (n1 - 1) = {sum_sq_diff_cluster1:.2f} / ({n1} - 1) = {variance_cluster1:.2f}")
        print(f"Standard Deviation (σ1) = sqrt(Variance) = sqrt({variance_cluster1:.2f}) = {std_dev_cluster1:.2f}")
    else:
        std_dev_cluster1 = 0.0 # Or np.nan, depending on how you want to handle single-point clusters
        print("Cluster 1 has 1 or fewer points, standard deviation is considered 0 or undefined.")
else:
    std_dev_cluster1 = None
    print("Cluster 1 is empty or its mean is not defined, cannot calculate standard deviation.")


# --- Standard Deviation Calculation for Cluster 2 (σ2) ---
print("\n--- Calculations for Cluster 2 Standard Deviation (σ2) ---")
if current_cluster2_points and current_mean_cluster2 is not None:
    n2 = len(current_cluster2_points)
    if n2 > 1: # Standard deviation requires at least 2 points
        # Calculate sum of squared differences from the mean
        sum_sq_diff_cluster2 = sum([(x - current_mean_cluster2)**2 for x in current_cluster2_points])
        variance_cluster2 = sum_sq_diff_cluster2 / (n2 - 1) # Sample variance (n-1 degrees of freedom)
        std_dev_cluster2 = math.sqrt(variance_cluster2)

        print(f"Sum of (x - µ2)^2 for Cluster 2:")
        for x_val in current_cluster2_points:
            diff_sq = (x_val - current_mean_cluster2)**2
            print(f"  ({x_val:.1f} - {current_mean_cluster2:.2f})^2 = {diff_sq:.2f}")
        print(f"  Sum = {sum_sq_diff_cluster2:.2f}")

        print(f"Number of points in Cluster 2 (n2): {n2}")
        print(f"Variance (σ2^2) = Sum((x - µ2)^2) / (n2 - 1) = {sum_sq_diff_cluster2:.2f} / ({n2} - 1) = {variance_cluster2:.2f}")
        print(f"Standard Deviation (σ2) = sqrt(Variance) = sqrt({variance_cluster2:.2f}) = {std_dev_cluster2:.2f}")
    else:
        std_dev_cluster2 = 0.0 # Or np.nan
        print("Cluster 2 has 1 or fewer points, standard deviation is considered 0 or undefined.")
else:
    std_dev_cluster2 = None
    print("Cluster 2 is empty or its mean is not defined, cannot calculate standard deviation.")

print(f"\n--- Final Standard Deviation Results ---")
print(f"Standard Deviation of Cluster 1 (σ1): {std_dev_cluster1:.2f}" if std_dev_cluster1 is not None else "N/A")
print(f"Standard Deviation of Cluster 2 (σ2): {std_dev_cluster2:.2f}" if std_dev_cluster2 is not None else "N/A")
```
